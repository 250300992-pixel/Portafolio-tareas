```python
#P-2.36 Escriba un programa en Python para simular un ecosistema con dos tipos de criaturas:
#osos y peces. El ecosistema consiste en un río, modelado como una lista relativamente grande.
#Cada elemento de la lista debe ser un objeto Oso, un objeto Pez o Ninguno. En cada paso de tiempo,
#según un proceso aleatorio, cada animal intenta moverse a una ubicación adyacente de la lista o
#permanecer donde está. Si dos animales del mismo tipo están a punto de colisionar en la misma celda,
#permanecen donde están, pero crean una nueva instancia de ese tipo de animal, que se coloca en una
#ubicación vacía aleatoria (es decir, previamente Ninguno) en la lista. Sin embargo, si un oso y un
#pez colisionan, el pez muere (es decir, desaparece).
```

```python
import random

class Animal:
    def __init__(self):
        pass

class Bear(Animal):
    def __str__(self):
        return "🐻"

class Fish(Animal):
    def __str__(self):
        return "🐟"

class River:
    def __init__(self, size=20):
        self.size = size
        self.cells = [None] * size


        for i in range(size):
            r = random.random()
            if r < 0.2:
                self.cells[i] = Bear()
            elif r < 0.5:
                self.cells[i] = Fish()


    def display(self):
      print([str(c) if c else "·" for c in self.cells])

River().display()
```
